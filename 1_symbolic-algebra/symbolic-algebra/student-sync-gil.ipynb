## I.C. Variable Domain

::::{tip}🐍 Specifying the Domain of a SymPy Variable
:icon:false
When a SymPy variable represents a chemically or physically observable quantity, it is important to define the _variable domain_ to ensure SymPy behaves in a physically reasonable manner. To specify the domain of a SymPy variable, set one or more _assumptions_ to be `True` by passing them as keyword arguments to the `var()` function:

```{code}python
:linenos:
:filename:Syntax Demonstration
# Variables representing arbitrary real numbers
var('x y z', real=True)
# Positive, nonzero real number representing mass
var('m', real=True, positive=True, nonzero=True)
# Nonnegative integer representing principle quantum number
var('n', integer=True, nonnegative=True)
```

:::{warning} ⚠️ Caution: Separate declaration of variables with different domains or assumptions
:icon:false
:class:dropdown
The keyword arguments passed to `var()` to specify domain and any other assumptions apply to _every_ variable declared with that call to `var()`, so if different variables have different domains/assumptions, they must be declared separately! 

For example, in thermochemical contexts the heat $q$ can be negative, zero, or positive, but the absolute temperature $T$ can only be positive:
```{code}python
:linenos:
:filename:Syntax Demonstration
# Declare variables
var('q', real=True) # Heat can be any real number
var('m C T_1 T_2', real=True, positive=True, nonzero=True) # mass, heat capacity, and temperature must strictly positive
```
:::

Below are listed several variable domains which may be appropriate depending on the chemical context or application of interest, as well as several other useful assumptions that can be used to restrict which values in the larger domain are allowed.

:::{note}ℹ️ Common Variable Domains
:icon:false
:class:dropdown
Definitions for some potentially relevant variable domains, as well as how they may be specified as a keyword argument to the `var()` function.

Real Numbers
: A _real number_ $c$ is any numerical value $-\infty < c < \infty$ that can be measured; the set of all real numbers is denoted $\mathbb{R}$.
    * {u}`Examples:` $5$, $\pi$, $5.32$, $\tfrac{1}{2}$
    * {u}`Syntax:` `real=True`

Imaginary Numbers
: _**The** imaginary number_ is $i = \sqrt{-1}$. 
: A number $z$ is said to be _imaginary_ (a.k.a. _pure imaginary_) if it is a multiple of the imaginary number $i$, i.e., $z = \pm bi$ for some real number $b$.
    * {u}`Syntax:` `imaginary=True`

Complex Numbers
: A _complex number_ $z = a \pm bi$ (for real constants $a,b$) is one which contains _both_ a real and imaginary part; the set of all complex numbers is denoted $\mathbb{C}$.
    * {u}`Syntax:` `complex=True`

Integers
: An _integer_ is a real counting number, i.e., $0, +1, -1, +2, -2,\ldots$; the set of all integers is denoted $\mathbb{Z}$.
    * {u}`Syntax:` `integer=True`
:::
:::{note}ℹ️ Additional Useful Assumptions
:icon:false
:class:dropdown
For the purposes of illustration, let $x$ represent a real number. The following assumptions can be passed as keyword arguments to `var()` to add additional limitations on which values within $\mathbb{R}$ (or any other domain) the variable $x$ is allowed to assume. 

Positive
: $x > 0$

Negative
: $x < 0$

Nonnegative
: $x\geq 0$

Nonpositive
: $x\leq 0$

Nonzero
: $x\neq 0$
:::

For a complete list of variable domains and possible assumptions, along with precise mathematical definitions for each, refer to {u}`[the Predicates section of the SymPy documentation.](https://docs.sympy.org/latest/guides/assumptions.html#predicates)`
::::

:::{danger}☢️ Danger: Variables with Unspecified Domain May Behave Unphysically
:icon:false
SymPy is a general purpose mathematics library, and by default assumes that all variables are complex (i.e., a member of $\mathbb{C}$ with both a real and imaginary part). This can lead to potentially unphysical behavior when solving equations or performing calculus operations like integration, which will produce erroneous results even if no explicit syntax error is thrown.

To avoid unphysical behavior, ***explicitly specify the domain of all SymPy variables when they are declared!***
:::

# II. Evaluating a SymPy Expression

## II.A. Substituting Values into an Expression

::::{tip}🐍 Substituting Values into a SymPy Expression
:icon:false
To evaluate a SymPy expression for particular values of its variables, use the `.subs()` method, whose argument is a Python dictionary mapping each SymPy variable to a particular value.

:::{note}👀 Python Dictionaries
:icon: false
:class:dropdown
Python dictionaries are objects of type `dict` which associate values with the keys that are used to "look them up," and are defined by placing comma-separated `key: value` pairs inside curly braces like the following:
```{code}python
new_dict = {'key_1': 'value 1', 'key_2': 'value 2', 'key_3': 3, 'key_4': ideal_T}
```
In general, a dictionary's keys should be defined as strings (type `str`), and behave like Python variables that only live inside the dictionary. The dictionary's values, on the other hand, can be any mixture of types (i.e., not all entries in the dictionary must be of the same type). To access the value associated with a particular key in the dictionary, you would use square brackets `[]`:

```{code}python
:linenos:
:filename: Cell Input
print(new_dict[key_1])
```
```{code}txt
:filename: Cell Output
value_1
```
This is similar to accessing a member of a `list`, except instead of passing the integer index of the list element, you pass the `str` name of the desired value's key inside the brackets.
:::

For example, the following syntax will evaluate the function $f(x) = -ax^2$ where $-\infty < x < \infty$ and $a>0$ for the particular values $a=\tfrac{1}{2}$ and $x=1.234$:
```{code}python
:linenos:
:filename:Syntax Demonstration: Evaluating a SymPy Expression with subs()
# Declare variables, define function f(x)
var('x', real=True)
var('a', real=True, positive=True, nonzero=True)
f = exp(-a * x**2)

# Substitute values with subs(), directly pretty-print output
f.subs({x:1.234, a:1/2})
```
:::{note}⏎ Cell Output
:icon:false
$\displaystyle 0.467022426491753$
:::

# III. Mathematical Equalities

## III.A. Defining Named Equalities

::::{tip}🐍 Encoding a Mathematical Equality into SymPy
:icon:false
To encode a mathematical equality in SymPy and assign it to a Python variable so that it can be manipulated later, use the special assignment operator `=@`.

For example, the equation defining a circle in the $xy$-plane centered at the origin with radius $r=1$ is
\begin{equation*}
x^2 + y^2 = 1
\end{equation*}
which can be encoded and assigned to the Python variable `unit_circle` with the syntax
```{code}python
:linenos:
:filename:Syntax Demonstration: Defining & Assigning a Mathematical Equality into SymPy
# Declare variables for unit circle
var('x y', real=True)
# Define ideal gas law
unit_circle =@ x**2 + y**2 = 1
# Pretty-print the ideal gas law equation from SymPy
unit_circle
```
:::{note} ⏎ Cell Output
:icon:false
$\displaystyle x^2 + y^2 = 1\;\;\; ({\tt unit\_circle})$
:::

:::{note}👀 Note: Mathematical $=$ vs. Python Assignment Operator `=`
:icon:false
:class:dropdown
Equations involve two expressions which are mathematically equivalent, which is represented symbolically using the _mathematical equals sign_ "$=$", i.e., 
\begin{equation*}
{\rm left-hand\ side\ (lhs)} = {\rm right-hand\ side\ (rhs)}
\end{equation*}
Unfortunately, this symbol has a double-meaning in Python, as the _assignment oeprator_ `=` which uniquely associates a value with a named variable, i.e., 
```{code}python
my_var = 5
```
To distinguish these contexts, this and future lessons will specify either the _mathematical equals_ "$=$" vs. the _Python assignment operator_ `=`.
:::
:::{danger}⛔️ Error Avoidance: Do Not Include Special Assignment Operator `=@` in Code Comments
:icon:false
:class:dropdown
The special assignment operator `=@` is not standard Python, but is instead a custom addition of the `algebra_with_sympy` wrapper library that we use to make SymPy easier to use. Because of how the `=@` operator is captured by `algebra_with_sympy`, this operator _can ONLY be present when syntactically meaningful._ ***Even having `=@` inside a code comment will throw a syntax error!!!***

## III.B. Manual Algebraic Rearrangement

::::{tip}🐍 Manual Algebraic Rearrangement of a SymPy Equation
:icon:false
Consider the equation $x + 5 = 2$. To algebraically rearrange this equation to solve for $x$, we need to subtract 5 from both sides:
\begin{align*}
x + &5 &&= 2 \\
  - &5 &&{}\;\;\;\; - 5\\
    &x &&= 2 - 5 = -3
\end{align*}
In SymPy all arithmetic operations performed on a named equation will be applied to both sides of the equality, so the rearrangement above can be accomplished with
```{code}python
:linenos:
:filename:Syntax Demonstration
# Declare variables & define named equation
var('x', real=True)
eqn =@ x + 5 = 2
# Rearrange by subtracting 5 from each side, assign to Python variable for posterity
eqn_x = eqn - 5
# Pretty-print w/ cell return
eqn_x
```
:::{note} ⏎ Cell Return
:icon:false
$\displaystyle x = -3$
:::
:::{warning}⚠️ Author Note: Assignment Statements Not Involving Math "$=$"
:icon:false
All Python assignment statements _which do not explicitly include the mathematical equals character_ $=$ in the SymPy equation must use the standard Python assignment operator `=` and **not the special assignment character** `=@`!
:::

## IV.B. Units in SymPy Expressions

::::{tip}🐍 Declaring Units with `units()`
:icon:false
Before units can be used in SymPy, they must first be declared using the `units()` function in an identical manner to declaring a SymPy variable with the `var()` function.  

For example, to declare units of meters, seconds, and kilograms:
```{code}python
:linenos:
:filename:Syntax for the units() function
units('m s kg')
```
Once units have been declared, they can be substituted into a previously defined SymPy expression and used algebraically to perform dimensional analysis, however units in some expressions may not completely simplify.
:::{warning}⚠️ Author Caution: Unit Relationships & Conversions
:icon:false
:class:dropdown
Units defined using the `units()` function are unaware of any relationships between them, either by way of metric prefix, definition, or conversion factor. This represents a major difference between how we might expect declared units to behave and how they _actually_ behave. Defining and using units for dimensional analysis is therefore almost always more trouble than it is worth.[^fn:algwsym-units]
:::
:::{danger}⛔️ Author Warning: NEVER Substitute Values with Units into Expressions
:icon:false
:class:dropdown
While it is technically possible to substitute numerical values _with_ associated units defined with the `units()` function into a SymPy expression using the `algebra_with_sympy` wrapper library, ***using this functionality is NOT recommended*** as it will break the SymPy `solve()` function and because of the challenges of incomplete unit simplification warned about above.
:::
[^fn:algwsym-units]: This is the opinion of the lesson author, based on three years of experience using the `algebra-with-sympy` wrapper library for both his personal purposes and when teaching these skills in his Physical Chemistry courses. The only significant use case he has found for this functionality is the extremely complicated dimensional analysis that arises in the context of deriving anharmonic rovibrational spectroscopic constants from quantum chemically constructed potential energy surfaces.

## IV.C. Logarithms in Python & SymPy

::::{danger}☢️ Danger: Logarithms in Python
:icon:false
Unlike in a normal mathematical context, where "$\log$" is understood to represent the common logarithm and "$\ln$" is understood to mean the natural logarithm, Python and many other programming languages actually use `log` to mean the natural logarithm and `log10` to mean the common logarithm. 

Because this and future lessons involve both the numerical evaluation and symbolic manipulation of the natural & common logarithms, the logarithm functions from two separate libraries have been imported into this notebook in the {u}`[Package Imports](#package-imports)` section, summarized below:

| Characteristic                                                        | Natural Logarithm | Common Logarithm |
|:----------------------------------------------------------------------|:-----------------:|:----------------:|
